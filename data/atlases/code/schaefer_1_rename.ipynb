```python
import os, sys, shutil
import numpy as np
import pandas as sp
import pandas as pd
import nibabel as nib
import json
import wget
```

```python
home = os.path.expanduser("~")
print(home)
atlas_dir = os.path.join(home, 'research_projects/snaplab_tools/data/atlases')
```

## MNI

```python
def process_tsv(tsv_file):
    df = pd.read_csv(tsv_file, header=0, index_col=0, sep='\t')
    # df.index.name = 'index'
    df.rename(columns={'name': 'label'}, inplace=True)
    # df.drop(columns=['R', 'A', 'S'], inplace=True)
    df.drop(columns=['color'], inplace=True)
    
    df.to_csv(tsv_file, sep="\t")
    # print(df)
```

```python
# MNI152NLin2009cAsym
in_dir = os.path.join(home, 'templateflow/tpl-MNI152NLin2009cAsym')
for n_regions in [100, 200, 400]:
    out_dir = os.path.join(atlas_dir, 'Schaefer/atlas-Schaefer{0}7'.format(n_regions))
    if os.path.isdir(out_dir):
        shutil.rmtree(out_dir)
    os.makedirs(out_dir)
    for res in [1, 2]:
        shutil.copyfile(
            os.path.join(in_dir, 'tpl-MNI152NLin2009cAsym_res-0{0}_atlas-Schaefer2018_desc-{1}Parcels7Networks_dseg.nii.gz'.format(res, n_regions)),
            os.path.join(out_dir, 'atlas-Schaefer{0}7_space-MNI152NLin2009cAsym_res-0{1}_dseg.nii.gz'.format(n_regions, res))
            )

    # text file
    tsv_file = os.path.join(out_dir, 'atlas-Schaefer{0}7_dseg.tsv'.format(n_regions))
    shutil.copyfile(
        os.path.join(in_dir, 'tpl-MNI152NLin2009cAsym_atlas-Schaefer2018_desc-{0}Parcels7Networks_dseg.tsv'.format(n_regions)),
        os.path.join(tsv_file)
        )
    process_tsv(tsv_file)

    json_file = os.path.join(out_dir, 'atlas-Schaefer{0}7_dseg.json'.format(n_regions))
    data = {"BIDSVersion": "1.8.0", "Name": "Schaefer{0}7".format(n_regions)}
    # creating a JSON string
    json_string = json.dumps(data)
    # storing it in a file
    with open(json_file, "w") as json_data:
        json.dump(data, json_data)

# MNI152NLin6Asym
in_dir = os.path.join(home, 'templateflow/tpl-MNI152NLin6Asym')
for n_regions in [100, 200, 400]:
    out_dir = os.path.join(atlas_dir, 'Schaefer/atlas-Schaefer{0}7'.format(n_regions))
    for res in [1, 2]:
        shutil.copyfile(
            os.path.join(in_dir, 'tpl-MNI152NLin6Asym_res-0{0}_atlas-Schaefer2018_desc-{1}Parcels7Networks_dseg.nii.gz'.format(res, n_regions)),
            os.path.join(out_dir, 'atlas-Schaefer{0}7_space-MNI152NLin6Asym_res-0{1}_dseg.nii.gz'.format(n_regions, res))
            )
```
